```python
import numpy as np
import skfuzzy as fuzz
from skfuzzy import control as ctrl
```

### ctrl.Antecedent & ctrl.Consequent

Ini adalah wadah atau "variabel" untuk menampung rentang angka yang sudah kita buat di atas. Antecedent untuk Input (Kondisi/Sebab), Consequent untuk Output (Hasil/Akibat).

```python
suhu = ctrl.Antecedent(np.arange(0,46), 'suhu')
kipas = ctrl.Consequent(np.arange(0,101), 'kecepatan')
```

### Fungsi Keanggotaan: .automf dan fuzz.trimf

Ini adalah proses Fuzzifikasi. Kita membagi angka-angka tadi menjadi kata-kata (kategori).

```python
suhu.automf(names=['dingin', 'normal', 'panas'])
```

```python
kipas['lambat'] = fuzz.trimf(kipas.universe, [0, 0, 50])
kipas['cepat'] = fuzz.trimf(kipas.universe, [50, 100, 100])

print(kipas['lambat'].mf)
```

### ctrl.Rule

Di sinilah letak "kepakaran" sistem Anda. Anda menghubungkan Input dan Output dengan logika JIKA-MAKA.

```python
rule1 = ctrl.Rule(suhu['panas'], kipas['cepat'])
rule2 = ctrl.Rule(suhu['dingin'], kipas['lambat'])
rule1
```

```python
mesin = ctrl.ControlSystem([rule1, rule2])
simulation = ctrl.ControlSystemSimulation(mesin)
```

### ControlSystem & Eksekusi (.input, .compute(), .output)

Ini adalah tahap akhir. Kita bungkus aturannya, buat mesinnya, masukkan angka sungguhan, lalu minta hasil akhirnya (Defuzzifikasi).

```python
simulation.input['suhu'] = 36
simulation.compute()

print(simulation.output)
```
